# Initial Analysis of UFC Master Dataset

**Firstly Collecting Raw UFC Data:**

```python
from pathlib import Path
import pandas as pd

# Always calculate relative to notebook location
NOTEBOOK_DIR = Path.cwd()  # This is /.../ml_core/notebooks/
PROJECT_ROOT = NOTEBOOK_DIR.parent.parent  # Go up 2 levels: /.../ufc-fight-predictor/

# Build the data path to where raw data source is located (ufc-master.csv)
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "ufc-master.csv"
print(f"📁 Project root: {PROJECT_ROOT}")
print(f"📄 Data path: {DATA_PATH}")

# Load data
df = pd.read_csv(DATA_PATH)
print(f"✅ Loaded {len(df)} rows")
```

```python
# displaying sample of dataset to check for early necessary changes to be made

pd.set_option('display.max_columns', None) # Want to view columns to possibly be used as features

# Encode categorical stuff to numbers so that it can be enabled as features

display(df.head(3)) # check for first few rows only
```

First, will remove columns that represent information of what happened during the fight:
- Finish
- FinishDetails
- FinishRound
- FinishRoundTime
- TotalFightTimeSecs

```python
df = df.drop(columns=["Finish", "FinishDetails", "FinishRound", "FinishRoundTime", "TotalFightTimeSecs"])
```

Now to prepare the ufc-master training dataset before being fed into the ML model, which would require converting all columns (x features) with string values into numerical values via an encoder, and removing unwanted columns/features for training:

List of all columns that contain string values that will be converted into numerical values for training:
- Winner (this is our y feature) (Fights that resulted in NC or draws have already been removed from raw training dataset)
- TitleBout (True, False)
- WeightClass
- Gender (MALE, FEMALE)
- BlueStance, RedStance
- BetterRank (neither, Red, Blue)

**Will now encode using oridnal encoding on the columns based on a ranking system and the columns that can be represented as a binary state:**
- This would include TitleBout, WeightClass, Gender, and BetterRank

```python
from sklearn.preprocessing import OrdinalEncoder

# encoding Winner column to numerical values (Fights that didn't result in a winner have already been removed from training data)
encoded_order = [["Red", "Blue"]]
ordinal_encoder = OrdinalEncoder(categories=encoded_order)

# applying encoded values to dataframe as separate column
df['Winner_Encoded'] = ordinal_encoder.fit_transform(df[["Winner"]]).astype(int)


# encoding TitleBout column to numerical values
encoded_order = [[False, True]]
ordinal_encoder = OrdinalEncoder(categories=encoded_order)

# applying encoded values to dataframe as separate column
df['TitleBout_Encoded'] = ordinal_encoder.fit_transform(df[["TitleBout"]]).astype(int)


# encoding WeightClass column to numerical values, from lightest to heaviest for men's weight classes, then women's weight classes, then catchweight
encoded_order = [["Flyweight", "Bantamweight", "Featherweight", "Lightweight", "Welterweight", "Middleweight", 
                  "Light Heavyweight", "Heavyweight", "Women's Strawweight", "Women's Flyweight", "Women's Bantamweight", 
                  "Women's Featherweight", "Catch Weight"]]
ordinal_encoder = OrdinalEncoder(categories=encoded_order)

# applying encoded values to dataframe as separate column
df['WeightClass_Encoded'] = ordinal_encoder.fit_transform(df[["WeightClass"]]).astype(int)


# encoding Gender column to numerical values
encoded_order = [["MALE", "FEMALE"]]
ordinal_encoder = OrdinalEncoder(categories=encoded_order)

# applying encoded values to dataframe as separate column
df['Gender_Encoded'] = ordinal_encoder.fit_transform(df[["Gender"]]).astype(int)


# encoding BetterRank column to numerical values
encoded_order = [["neither", "Red", "Blue"]]
ordinal_encoder = OrdinalEncoder(categories=encoded_order)

# applying encoded values to dataframe as separate column
df['BetterRank_Encoded'] = ordinal_encoder.fit_transform(df[["BetterRank"]]).astype(int)
```

**Now the rest of the columns that are encoded will not be based on a 'rank' or 'binary' pattern as previous columns, so one-hot encoder will now be used on the following columns:**
- BlueStance
- RedStance
---
**Will now define a function to perform one-hot encoding and dynamically create x-1 number of new columns based on x distinct values from the original string column**

```python
from sklearn.preprocessing import OneHotEncoder

"""
Dynamically one-hot encode specified column to create multiple new encoded columns.
    
Parameters:
    df (pd.DataFrame): Input DataFrame
    column_to_encode (string): List of column names to encode
    
Returns:
    pd.DataFrame: Fully encoded DataFrame
"""
def one_hot_encode(df_uncoded, column_to_encode):

    encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

    # Fit and transform the target column
    encoded = encoder.fit_transform(df_uncoded[[column_to_encode]])

    # Create a new dataframe with all the new column names to be added
    df_encoded = pd.DataFrame(
        encoded,
        columns=encoder.get_feature_names_out([column_to_encode]),
        index=df_uncoded.index
    )

    # Now add the new columns to the original, uncoded version of the df
    df_final = pd.concat([df_uncoded, df_encoded], axis=1)

    return df_final
```

**Now encoding the mentioned columns using One-Hot Encoder:**

```python
# Using one-hot encoding on BlueStance
df = one_hot_encode(df, "BlueStance")

# On RedStance
df = one_hot_encode(df, "RedStance")
```

**Now that everything is encoded, we can remove all the string columns from our training dataset:**
- Location
- Country
- EmptyArena
- Winner
- TitleBout
- WeightClass
- Gender
- BlueStance
- RedStance
- BetterRank

**RedFighter, BlueFighter and Date are deliberately kept, as we will need to know which fighter this information belongs to when trying to predict upcoming mathcups of specific fighters and we will need the information based on the date of their latest fight. These labels will be temporarily removed when training our models.**

```python
# Dropping unwanted columns for training
df = df.drop(columns=["Location", "Country", "EmptyArena", "Winner", "TitleBout", 
                      "WeightClass", "Gender", "BlueStance", "RedStance", "BetterRank"])
```

- **Will Now remove all null values from our x features, in particular the columns that contain a fighters a rank at a particular weight class.** 
- **If a fighter is unranked and the value for that column is null, 999 value will replace that null instead to make the fighter being unranked for that particular weight class meaningful in our training data**

```python
# Columns containing info about a fighters rank, from which the null values need to be removed
rank_columns = ["BMatchWCRank", "RMatchWCRank", "RWFlyweightRank", "RWFeatherweightRank", "RWStrawweightRank", "RWBantamweightRank", "RHeavyweightRank", 
                "RLightHeavyweightRank", "RMiddleweightRank", "RWelterweightRank", "RLightweightRank", "RFeatherweightRank", "RBantamweightRank", 
                "RFlyweightRank", "RPFPRank", "BWFlyweightRank", "BWFeatherweightRank", "BWStrawweightRank", "BWBantamweightRank", "BHeavyweightRank", 
                "BLightHeavyweightRank", "BMiddleweightRank", "BWelterweightRank", "BLightweightRank", "BFeatherweightRank", "BBantamweightRank", 
                "BFlyweightRank", "BPFPRank"]

# replace all null's with 999
for col in rank_columns:
    df[col] = df[col].fillna(999)


# Will now check for other columns with nulls to be dealt with:
    
from sklearn.impute import SimpleImputer

# Will use simple imputer to replace nulls found within the rest of columns with the average value of that column
imputer = SimpleImputer(strategy='mean')

# Grabbing the total null count of each column in our training data
null_counts = df.isnull().sum()

# Only columns with nulls found
null_counts = null_counts[null_counts > 0]

# Now removing the nulls from these columns and replacing it with the average
columns_nulls = null_counts.index.tolist()
df[columns_nulls] = imputer.fit_transform(df[columns_nulls])
```
