# Saving Random Forest Classifier as Pre-Trained Model on Full Training Data

**Load our full training data:**

```python
from pathlib import Path
import pandas as pd


# Always calculate relative to notebook location
NOTEBOOK_DIR = Path.cwd()  # This is /.../ml_core/notebooks/
PROJECT_ROOT = NOTEBOOK_DIR.parent.parent  # Go up 2 levels: /.../ufc-fight-predictor/

# Build the data path to where processed training data source is located (ufc-master-processed.csv)
DATA_PATH = PROJECT_ROOT / "data" / "processed" / "ufc-master-processed.csv"
print(f"📁 Project root: {PROJECT_ROOT}")
print(f"📄 Data path: {DATA_PATH}")

# Load data
df = pd.read_csv(DATA_PATH).drop(columns=["RedFighter", "BlueFighter", "Date"])

# Separating our X and y features
X = df.drop(columns=['Winner_Encoded']) # All our X features
y = df['Winner_Encoded'] # Our target


# Verifying size of original data, training set and test set
print("Total number of rows:")
print(f"Original data: {len(df)}")
```

**Train Random Forest Classifier on full training data instead of only the training split:**

```python
from sklearn.ensemble import RandomForestClassifier
import joblib
from pathlib import Path

# Load our model
model = RandomForestClassifier(random_state=42, n_estimators=100)

# Train on full list of X features and y feature
model.fit(X, y)

# Always calculate relative to notebook location
NOTEBOOK_DIR = Path.cwd()  # This is saving to /.../ml_core/models/
PROJECT_ROOT = NOTEBOOK_DIR.parent.parent  # Go up 2 levels: /.../ufc-fight-predictor/

# Build the data path to where processed training data source is located (ufc-master-processed.csv)
MODEL_PATH = PROJECT_ROOT / "ml_core" / "models" / "trained_model.joblib"
COLUMNS_PATH = PROJECT_ROOT / "ml_core" / "models" / "feature_columns.joblib"
print(f"📄 Model path: {MODEL_PATH}")
print(f"📄 Columns path: {COLUMNS_PATH}")

MODEL_PATH.parent.mkdir(parents=True, exist_ok=True) # in case models/ doesn't exist yet

joblib.dump(model, filename=MODEL_PATH) # save trained model to specified location
joblib.dump(list(X.columns), filename=COLUMNS_PATH) # save the exact column order the model expects
```
